# Optimization Lab 1

## 1 One dimensional case

We plot $f(x) = x^3 - 2x + 2$

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x*x*x-2*x-2

x = np.linspace(-2, 2, 400)
y = f(x)

plt.figure(figsize=(6, 4))
plt.plot(x, y, label=r"$f(x) = x^3 - 2x + 2$", color="blue")
plt.axhline(0, color='black', linewidth=0.8, linestyle="--")
plt.axvline(0, color='black', linewidth=0.8, linestyle="--")
plt.title("Plot of $f(x) = x^3 - 2x + 2$")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.grid(True)
plt.show()
```

The points that satisfy $f'(x) = 3x^2 - 2 = 0$ are $x_1 = \sqrt\frac{3}{2}$ and $x_2 = -\sqrt\frac{3}{2}$ that are congruent with the plot

We now need to see if $f''(x)$ is positive or negative. In fact $f''(x) = 6x$ so it will be positive for $x_1$ therefore a maximum and negative for $x_2$ therefore a minimum.
We can plot it:

```python
def f2(x):
    return 6*x

x = np.linspace(-2, 2, 400)
y = f2(x)

plt.figure(figsize=(6, 4))
plt.plot(x, y, label=r"$f''(x) = 6x$", color="blue")
plt.axhline(0, color='black', linewidth=0.8, linestyle="--")
plt.axvline(0, color='black', linewidth=0.8, linestyle="--")
plt.title("Plot of $f''(x) = 6x$")
plt.xlabel("x")
plt.ylabel("$f''(x)$")
plt.legend()
plt.grid(True)
plt.show()
```

## 2.2 A two dimensional function with multiple minima

We plot the function $f(x_1, x_2) = x_1^2 \left( 4 - 2.1x_1^2 + \tfrac{1}{3}x_1^4 \right) + x_1x_2 + x_2^2 \left( -4 + 4x_2^2 \right)$

```python
from mpl_toolkits.mplot3d import Axes3D 


def f(x1, x2):
    return (x1**2)*(4 - 2.1*(x1**2) + (1.0/3.0)*(x1**4)) + x1*x2 + (x2**2)*(-4 + 4*(x2**2))

# Define the grid
x1 = np.arange(-2.0, 2.0 + 0.1, 0.1)
x2 = np.arange(-1.0, 1.0 + 0.1, 0.1)
X1, X2 = np.meshgrid(x1, x2)
Z = f(X1, X2)

#3D Surface
fig1 = plt.figure(figsize=(8,6))
ax = fig1.add_subplot(111, projection='3d')
surf = ax.plot_surface(X1, X2, Z, cmap="viridis", edgecolor="none")
ax.set_title("Surface plot of f(x1, x2)")
ax.set_xlabel("x1")
ax.set_ylabel("x2")
ax.set_zlabel("f(x1, x2)")
plt.colorbar(surf, shrink=0.5, aspect=5)  # color bar for values
plt.show()

# Contour
plt.figure(figsize=(8,6))
contour = plt.contour(X1, X2, Z, levels=25, cmap="viridis")
plt.clabel(contour, inline=True, fontsize=8)
plt.title("Contour plot of f(x1, x2)")
plt.xlabel("x1")
plt.ylabel("x2")
plt.axvline(0, color='gray', linestyle='--', linewidth=0.8)
plt.axhline(0, color='gray', linestyle='--', linewidth=0.8)
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```
