# One dimensional case

We begin with the one dimensional case. Assume that $x \in \mathbb{R}$ and that  

$$
f(x) = x^3 - 2x + 2
$$

You are asked to follow the next steps:

1. **Plot this function** within the range $x \in [-2,2]$, for instance.  
   For that purpose use the `matplotlib` from Python using the examples included within this document.


```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x*x*x-2*x-2

x = np.linspace(-2, 2, 400)
y = f(x)

plt.figure(figsize=(6, 4))
plt.plot(x, y, label=r"$f(x) = x^3 - 2x + 2$", color="blue")
plt.axhline(0, color='black', linewidth=0.8, linestyle="--")
plt.axvline(0, color='black', linewidth=0.8, linestyle="--")
plt.title("Plot of $f(x) = x^3 - 2x + 2$")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.grid(True)
plt.show()
```

2. **Compute analytically** the points $x^*$ that satisfy $f'(x) = 0$. Observe if the obtained result is congruent with the plot of the previous point.



The points that satisfy $f'(x) = 3x^2 - 2 = 0$ are $x_1 = \sqrt\frac{3}{2}$ and $x_2 = -\sqrt\frac{3}{2}$ that are congruent with the plot.

3. We are now going to **check which of the latter points $x^*$ are minima (or a maxima)**.  
   For that purpose let us perform a 2nd order Taylor expansion around point \(x^*\):

   $$
   f(x^* + d) \approx f(x^*) + d f'(x^*) + \tfrac{1}{2} d^2 f''(x^*)
   $$

   where $d \in \mathbb{R}$ is the perturbation around $x^*$. Since we are dealing with a one dimensional function, $f''(x^*)$ is a real number which may be positive or negative.


We now need to see if $f''(x)$ is positive or negative. In fact $f''(x) = 6x$ so it will be positive for $x_1$ therefore a maximum and negative for $x_2$ therefore a minimum.

4. You may also **plot $f''(x)$ within the range $x \in [-2,2]$**. If $f''(x)$ is positive, the function may be approximated with a convex 2nd order Taylor expansion at $x$. On the other hand, if $f''(x)$ is negative the function is concave at that point.


```python
def f2(x):
    return 6*x

x = np.linspace(-2, 2, 400)
y = f2(x)

plt.figure(figsize=(6, 4))
plt.plot(x, y, label=r"$f''(x) = 6x$", color="blue")
plt.axhline(0, color='black', linewidth=0.8, linestyle="--")
plt.axvline(0, color='black', linewidth=0.8, linestyle="--")
plt.title("Plot of $f''(x) = 6x$")
plt.xlabel("x")
plt.ylabel("$f''(x)$")
plt.legend()
plt.grid(True)
plt.show()
```

## 2  Two dimensional case

### 2.1 A simple two-dimensional function

We are now going to focus on simple two-dimensional functions, $\mathbf{x} \in \mathbb{R}^2$, $\mathbf{x} = (x_1, x_2)^T$ (vectors are expressed column-wise). Let us begin with the next quadratic expression:

$$
f(\mathbf{x}) = x_1^2 + x_2^2
$$

Follow the next steps:

1. **Plot this function**. Please, note that this function has a minimum at $(x_1, x_2) = (0,0)$.

---

```python
import numpy as np
import matplotlib.pyplot as pltimport 
import plotly.graph_objects as go

R = 3.0        # rango para x1 y x2: [-R, R]
N = 301        # puntos por eje

x = np.linspace(-R, R, N)
y = np.linspace(-R, R, N)
X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

fig = plt.figure(figsize=(12, 5))

# Subplot 1: superfície 3D
ax1 = fig.add_subplot(1, 2, 1, projection='3d')
ax1.plot_surface(X, Y, Z, rstride=3, cstride=3, 
                 linewidth=0, antialiased=True, 
                 alpha=0.9, cmap="viridis")
ax1.set_title(r"$f(x_1,x_2)=x_1^2+x_2^2$ (superficie)")
ax1.set_xlabel(r"$x_1$")
ax1.set_ylabel(r"$x_2$")
ax1.set_zlabel(r"$f(x)$")

# Subplot 2: gràfic de contorn
ax2 = fig.add_subplot(1, 2, 2)
cs = ax2.contour(X, Y, Z, levels=15, cmap="viridis")
ax2.clabel(cs, inline=True, fontsize=8)
ax2.set_xlabel(r"$x_1$")
ax2.set_ylabel(r"$x_2$")
ax2.set_title(r"Contour plot of $f(x_1,x_2)=x_1^2+x_2^2$")
ax2.axhline(0, linewidth=0.8, color="k")
ax2.axvline(0, linewidth=0.8, color="k")
ax2.set_aspect("equal", adjustable="box")

plt.tight_layout()
plt.show()

```

---
2. Analytically **compute the gradient of the function**, $\nabla f(\mathbf{x})$, and compute the point $\mathbf{x}^*$ at which  
   $$
   \nabla f(\mathbf{x}^*) = 0.
   $$
---

So, firt of all we must compute de gradient of the function $f$, 

$$
\nabla f(\mathbf{x}) = (\dfrac{\partial f}{\partial x_1},\dfrac{\partial f}{\partial x_2}) = (2 x_1, 2 x_2) 
$$

and clearly, the point that cancels the gradient, $\nabla f(\mathbf{x}^*) = 0$, is $(x_1,x_2)=(0,0)$. 

---
3. Let $\mathbf{d} \in \mathbb{R}^2$ be the perturbation around $\mathbf{x}^*$. The Taylor expansion, up to second order, of a function of several variables can be compactly expressed as

   $$
   f(\mathbf{x}^* + \mathbf{d}) \approx f(\mathbf{x}^*) + \mathbf{d}^T \nabla f(\mathbf{x}^*) + \tfrac{1}{2} \mathbf{d}^T \nabla^2 f(\mathbf{x}^*) \mathbf{d}
   \tag{2}
   $$

Analyze the previous expression and be sure to understand the operations that are done at each of the terms.

Compute the Hessian matrix, $\nabla^2 f(\mathbf{x})$, at the point $\mathbf{x} = \mathbf{x}^*$. You should obtain
$$
\nabla^2 f(\mathbf{x}^*) =
\begin{pmatrix}
2 & 0 \\
0 & 2
\end{pmatrix}
$$

The latter matrix is giving us information about the shape of the quadratic approximation at $\mathbf{x} = \mathbf{x}^*$ in a similar way as has been done for the one dimensional case.

For the one-dimensional case it is easy to check if we have a minimum, $f''(x^*) > 0$, or a maximum, $f''(x^*) < 0$. For a higher dimensional problem we are sure that the quadratic approximation is convex and that we have a minimum if

$$
\mathbf{d}^T \nabla^2 f(\mathbf{x}^*) \mathbf{d} > 0 \quad \mathbf{d} \neq 0
\tag{3}
$$

We have a maximum if

$$
\mathbf{d}^T \nabla^2 f(\mathbf{x}^*) \mathbf{d} < 0 \quad \mathbf{d} \neq 0
\tag{4}
$$

The previous conditions can be verified by computing the eigenvalues of $\nabla^2 f(\mathbf{x}^*)$. If all eigenvalues are strictly positive, equation (3) is satisfied. If all eigenvalues are strictly negative, equation (4) is satisfied.  

For this example, which are the eigenvalues of the Hessian matrix? Do we have a minimum or a maximum at $\mathbf{x}^*$?

---

In this two dimentional case, the hessian matrix is given by:

$$
\nabla^2 f(\mathbf{x}) =
\begin{pmatrix}
\dfrac{\partial^2 f}{\partial x_1^2} & \dfrac{\partial^2 f}{\partial x_1 \partial x_2} \\
\dfrac{\partial^2 f}{\partial x_2 \partial x_1} & \dfrac{\partial^2 f}{\partial x_2^2}
\end{pmatrix} = 
\begin{pmatrix}
2 & 0 \\
0 & 2
\end{pmatrix}
\quad \forall \mathbf{x}\in \mathbb{R}^2
$$

clearly we can see that the eigenvalues of the hessian matrix are $\lambda_1$=$\lambda_2$=2. Then, because of all the eigenvalues are estrictly positive, we're in front a minimum value for $(0,0)$. Also, we can verify this results by performing the following expression (take into account that $\nabla^2 f(\mathbf{x})=2 \mathbb{I}_2 \quad \forall x\in\mathbb{R}^2$):

$$
d^T \nabla^2 f(\mathbf{x}^*) d=d^T 2 I_2 d = 2
\| d \| > 0 \quad \forall d\in \mathbb{R}^2
$$

**(b)** At the points where the gradient is zero, what kind of information is giving us the Hessian matrix?  
Is this a minimum? A maximum? None of both? You may use the `eigvals` function of Python to compute the eigenvalues of the Hessian matrix (i.e. there is no need to compute them analytically).

In a similar way as for the one dimensional case, the eigenvalues of the Hessian $\nabla^2 f(\mathbf{x})$ gives us information about the local shape of the function $f(\mathbf{x})$ at point $\mathbf{x}$. This information will be used by numerical methods to accelerate descent to the optimal point we are looking for! In this lab we will focus, however, on the optimal points.

---

Following the procedure of the last point, we compute de gradient for each functions, 

$$
\nabla f_A(\mathbf{x}) = (\dfrac{\partial f_A}{\partial x_1},\dfrac{\partial f_A}{\partial x_2}) = (-2 x_1, -2 x_2) 
$$

$$
\nabla f_B(\mathbf{x}) = (\dfrac{\partial f_B}{\partial x_1},\dfrac{\partial f_B}{\partial x_2}) = (2 x_1, -2 x_2) 
$$

$$
\nabla f_C(\mathbf{x}) = (\dfrac{\partial f_C}{\partial x_1},\dfrac{\partial f_C}{\partial x_2}) = (2 x_1, 0) 
$$


and clearly, the point that cancels all the gradient, $\nabla f_i(\mathbf{x}^*) = 0$ for $i \in \{A,B,C\}$, are $(x_1,x_2)=(0,0)$, specifically for $f_C$ all points $\{(0,y); y \in \mathbb{R}\}$ are critical points. Now, we obtain the hessian matrix for each function:

$$
\nabla^2 f_A(\mathbf{x}) = 
\begin{pmatrix}
-2 & 0 \\
 0 & -2
\end{pmatrix}
\quad \forall \mathbf{x}\in \mathbb{R}^2
$$

$$
\nabla^2 f_B(\mathbf{x}) = 
\begin{pmatrix}
 2 & 0 \\
 0 & -2
\end{pmatrix}
\quad \forall \mathbf{x}\in \mathbb{R}^2
$$

$$
\nabla^2 f_B(\mathbf{x}) = 
\begin{pmatrix}
 2 & 0 \\
 0 & 0
\end{pmatrix}
\quad \forall \mathbf{x}\in \mathbb{R}^2
$$

Then, for $f_A$ we can see that the eigenvalues are $\lambda_1 =\lambda_2 = -2$, so because of all the eigenvalues are negative (hessian matrix is strictly negative), we have a maximum point in $(0,0)$. For $f_B$, the eigenvalues are $\lambda_1 = 2$ and $\lambda_2 = -2$, that take different sign numbers, this mean we are in front a safflepoint. In the last one, $f_C$, the eigenvalues are $\lambda_1=2$ and $\lambda_1=0$, so we have a semidefinitive positive matrix, all the point in $x=0$ are global minimum.
